## Analysis

```python
'''
Group by
Aggregate Functions -> (mean, max, min, count, sum)
Order by -> sort_values (ascending= True/ False, by= column_name, ignore_index= True/ False (default))
rename -> dictionary {'current_column_name' : 'replaced_column_name'}
agg -> for analysis of multiple aggregate 
'''
```

```python
import pandas as pd
```

```python
df = pd.read_csv('hr_cleaned_data.csv')
```

```python
df.head()
```

```python
# Find the total employees hired through each Recruitment Source.
'''
    Syntax 
    -------------
    dataframe.groupby(column_name)[column_name].agg_function()
'''
# df['RecruitmentSource'].unique()
```

```python
emp_recruitment = df.groupby('RecruitmentSource')['EmpID'].count().reset_index()
emp_recruitment
```

```python
emp_recruitment = df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns = {'EmpID' : 'Total Employees', 'RecruitmentSource': 'RecruitmentSource'}
).sort_values(ascending= False, by='Total Employees', ignore_index=True)
emp_recruitment
```

```python
emp_recruitment = df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns = {'EmpID' : 'Total Employees', 'RecruitmentSource': 'RecruitmentSource'}
).sort_values(ascending= False, by='Total Employees', ignore_index=True)
emp_recruitment.tail()
```

```python
emp_recruitment = df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns = {'EmpID' : 'Total Employees', 'RecruitmentSource': 'RecruitmentSource'}
).sort_values(ascending= False, by='Total Employees', ignore_index=True)
emp_recruitment.head()
```

```python
# Find the total employees hired and total salary distribution of employees by each Recruitment Source and Gender.

rec_gen_sal = df.groupby(['RecruitmentSource', 'Gender']).agg(
    Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum')
).reset_index()
rec_gen_sal
```

```python

rec_gen_sal = df.groupby(['RecruitmentSource', 'Gender']).agg(
    Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Max_Salary = ('Salary', 'max'),
    Min_Salary = ('Salary', 'min')
).reset_index()
rec_gen_sal
```

```python
def changeCurrency(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}''{:,.2f}'.format)
```

```python

rec_gen_sal = df.groupby(['RecruitmentSource', 'Gender']).agg(
    Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Max_Salary = ('Salary', 'max'),
    Min_Salary = ('Salary', 'min')
).reset_index()
#rec_gen_sal['Total_Saalry'] = rec_gen_sal['Total_Salary'].map('${:,.2f}'.format)
changeCurrency(rec_gen_sal, 'Rs.', 'Total_Salary', 'Max_Salary', 'Min_Salary')
rec_gen_sal
```

```python
## Find total number of employees hired in each year.
emp_hired = df.groupby('HireYear')['EmpID'].count.reset_index().rename(
    columns = {'EmpID': 'Total Employees'}
)
#pct_change
emp_hired ['Hire%'] = round(emp_hired['Total Employees'].pct_change().fillna(0),2)*100
emp_hired['Hire%'] = emp_hired['Hire%'].map('{:,.2f}%'.format)
emp_hired
```

## Month wise total hire employees

```python
emp_month_hire = df.groupby(['HireMonth', 'HireMonth_Name'])['EmpID'].count().reset_index().rename(
    columns = {'EmpID': 'Total Employyes'}
)[['HireMonth_Name', 'Total Employees']]
emp_month_hire
```

## Day wise total Employees Hire

```python
days = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
df['HireDay_Name'] = pd.Categorical(df['HireDay_Name'], categories=days)
df['HireDay_Name'].unique()
```

```python
emp_day_hire = df.groupby('HireDay_Name', observed = False)['EmpID'].count().reset_index().rename(
    columns = {'EmpID' :'Total Employees'}
)
emp_day_hire
```
