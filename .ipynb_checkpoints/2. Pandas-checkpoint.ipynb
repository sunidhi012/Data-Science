```python
import pandas as pd
```

```python
# 1D -> Series
# 2D -> Dtaframe
```

```python
alphabets = pd.Series(['A', 'B', 'C', 'D', 'E'])
alphabets
```

```python
alphabets = pd.Series(['A', 'B', 'C', 'D', 'E', 'F'], index=['aa', 'bb', 'cc', 'dd', 'ee', 'ff'])
alphabets
```

```python
'''
Alias for Index
---------------
for columns:
columns = ['name1', 'name2', 'name3', '....', ..., 'name_n']

for rows:
index = ['name1', 'name2', 'name3', '....', ..., 'name_n']
```

```python

products = pd.DataFrame([
    ['Watch', 'NavyForce', 2018, 'Rs.35000'],
    ['TV', 'MI', 2022, 'Rs.45000'],
    ['Laptop', 'HP', 2024, 'Rs.150000'],
    ['Mobile', 'Samsung', 2024, 'Rs.150000']
], columns = ['Product Name', 'Brand', 'Model Year', 'Price'],
    index=['A', 'B', 'C', 'D'])
products
```

```python
alphabets[0:3]
```

## Slicing

### Loc and Iloc

```python
'''
loc -> Location
Iloc -> Integer Location

'''
```

```python
products.iloc[1:, 1:3]
```

```python
products.loc['A':'C', 'Brand':'Model Year']
```

```python
products.loc[
    ['A', 'C'],
    ['Brand', 'Model Year']
]
```

```python
products.loc[
    ['A', 'D'],
    ['Product Name', 'Price']
]
```

```python
products.iloc[
    [0, 2],
    [0, 2]
]
```

```python
products.iloc[
    [0, 3],
    [0, 3]
]
```

```python
products.iloc[:, 0]
```
