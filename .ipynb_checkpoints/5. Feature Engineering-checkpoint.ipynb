```python
def changeDateTime(dataframe, *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
```

```python
changeDateTime(df, 'DateofTermination', 'DOB', 'DateofHire', 'LastPerformanceReview_Date')
```

### Extract Year, Month, Day

```python
'''
year
month
day
'''
#df['DateofHire']
```

```python
df['HireYear'] = df['DateofHire'].dt.year
```

```python
df['HireMonth'] = df['DateofHire'].dt.month
df['HireDay'] = df['DateofHire'].dt.day
```

```python
## Extract date name
df['HireMonth_Name'] = df['DateofHire'].dt.month_name()
df['HireDay_Name'] = df['DateofHire'].dt.day_name()
```

## Convert Continuous data to Categorical data

```python
df['Salary'].min()
```

```python
df['Salary'].max()
```

```python
def convertSalary(Salary):
    if Salary > 0 and Salary <= 70000:
        return 'Low Salary'
    elif Salary > 70000 and Salary <= 150000:
        return 'Medium Salary'
    else:
        return 'High Salary'
```

```python
# apply
df['Salary_Label'] = df['Salary'].apply(convertSalary)
```

```python
df
```

## Feature Encoding

```python
'''
- Manual Encoding
    Dictionary
- Use map() function
    -> 

'''
```

```python
df['Salary_Label'].unique()
```

```python
# Label Encoder
#     High Salary - 0
#     Low Salary - 1
#     Medium Salary - 2
```

```python
salary_dict = {
    'Low Salary' : 0,
    'Medium Salary' : 1,
    'High Salary' : 2
}
```

```python
df['salary_label_num'] = df['Salary_Label'].map(salary_dict)
```

```python
def manualEncode(dataframe, col_name, *args):
    encode_value = {}
    for idx, val in enumerate(args):
        encode_value[val] = idx

    dataframe[f'{col_name}_id'] = dataframe[col_name].map(encode_value)
```

```python
manualEncode(df, 'PerformanceScore', 'Exceeds', 'Fully Meets', 'Needs Improvement', 'PIP')
```

```python
manualEncode(df, 'HireDay_Name', 'Tuesday', 'Monday', 'Thursday', 'Wednesday', 'Sunday', 'Friday',
       'Saturday')
```

### Split Columns

```python
'''
- Use of Split
-----------------------
- To seperate data into multiple columns on the basis of common character
- df['col_name'].str.split(find, n(no._of_columns) = index
'''
```

```python
hr = pd.read_csv('Dataset/HR_Dataset Refresh.csv')
hr.head()
```

```python
emp_name = hr['Employee_Name'].str.split(',', n=1, expand=True)
# emp_name.head()
```

```python
first_middle = emp_name[1].str.strip().str.split(' ', n=1, expand=True)
# first_middle.head()
```

```python
df['E_FirstName'] = first_middle[0]
df['E_MiddleName'] = first_middle[1]
df['E_LastName'] = emp_name[0]
df
```

```python
import numpy as np
```

```python
def splitName(name):
    value = name.strip().split()

    if len(value) == 1:
        return pd.Series([value[0], np.nan, np.nan])
    elif len(value) == 2:
        return pd.Series([value[0], np.nan, value[1]])
    else:
        return pd.Series([value[0], value[1], value[2]])
```
