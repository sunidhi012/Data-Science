```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import KMeans model
from sklearn.cluster import KMeans
# Import feature scaling packages
from sklearn.preprocessing import StandardScaler, LabelEncoder
# Import metrics
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('Dataset/Mall_Customers.csv')
df.head()
```

## Transfrom Gender Column

```python
le = LabelEncoder()
df['Gender_Id'] = le.fit_transform(df['Gender'])
```

```python
df.columns
```

## Features

```python
X = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Gender_Id']]
```

```python
scaler = StandardScaler()
X_scale = scaler.fit_transform(X)
# X_scale
```

## Elbow Method - KMeans

```python
k_range = range(1, 21)
error_rate = []

for i in k_range:
    kmeans = KMeans(n_clusters = i)
    kmeans.fit_predict(X_scale)
    error_rate.append(kmeans.inertia_)  # centroid point calculation
```

```python
plt.figure(figsize=(6, 4))
plt.plot(k_range, error_rate, marker='o', color='red')
plt.grid(linestyle='--', alpha=0.4)
plt.xlabel('K Range')
plt.ylabel('Inertia - Error Rate')
plt.title('Elbow Method to find Optimal Clusters - KMeans')
plt.show()
```

## Define Model

```python
k = 4
model = KMeans(n_clusters=k)
clusters = model.fit_predict(X_scale)
```

```python
df['cluster'] = clusters 
```

```python
sc = silhouette_score(X_scale, labels= clusters)
sc
```

```python
df.columns
```

```python
sns.scatterplot(data=df, x='Gender_Id', y='Annual Income (k$)', c = df['cluster'])
plt.show()
```
