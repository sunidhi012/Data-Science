## KPIs

## Total Loan Apllicants

```python
total_applicants = df['id'].count()
MTD_total_applicants = df[df['Issue_Month'] == 12]['id'].count()   # time series KPIs
MTD_total_applicants
```

```python
total_applicants = df['id'].count()
MTD_total_applicants = df[df['Issue_Month'] == 12]['id'].count()
PMTD_total_applicants = df[df['Issue_Month'] == 11]['id'].count()
MoM_total_applicants = round(((MTD_total_applicants - PMTD_total_applicants) / PMTD_total_applicants)* 100, 2)
```

```python
print(f'Total Applicants: {total_applicants}')
print(f'MTD Total Applicants: {MTD_total_applicants}')
print(f'PMTD Total Applicants: {PMTD_total_applicants}')
print(f'MoM Total Applicants: {MoM_total_applicants}%')
```

## Total Funded Amount

```python
funded_amount = df ['loan_amount'].sum()
MTD_funded_amount = df[df['Issue_Month'] == 12]['loan_amount'].sum()
PMTD_funded_amount = df[df['Issue_Month'] == 11]['loan_amount'].sum()
MoM_funded_amount = round(((MTD_funded_amount - PMTD_funded_amount) / PMTD_funded_amount)* 100, 2)
```

```python
print(f'Funded Amount: ${funded_amount:,.2f}')
print(f'MTD Funded Amount: ${MTD_funded_amount:,.2f}')
print(f'PMTD Funded Amount: ${PMTD_funded_amount:,.2f}')
print(f'MoM Funded Amount: {MoM_funded_amount}%')
```

## Total Received Amount

```python
received_amount = df ['total_payment'].sum()
MTD_received_amount = df[df['Issue_Month'] == 12]['total_payment'].sum()
PMTD_received_amount = df[df['Issue_Month'] == 11]['total_payment'].sum()
MoM_received_amount = round(((MTD_received_amount - PMTD_received_amount) / PMTD_received_amount)* 100, 2)
```

```python
print(f'Received Amount: ${received_amount:,.2f}')
print(f'MTD Received Amount: ${MTD_received_amount:,.2f}')
print(f'PMTD Received Amount: ${PMTD_received_amount:,.2f}')
print(f'MoM Received Amount: {MoM_received_amount}%')
```

```python
print(f'Average Intrest Rate: {int_rate:,.2f}%')
print(f'MTD Average Intrest Rate: {MTD_int_rate:,.2f}%')     # 0.123456040797403803
print(f'PMTD Average Intrest Rate: {PMTD_int_rate:,.2f}%')   # 0.1194171747211896
print(f'MoM Average Intrest Rate: {MoM_int_rate}%')
```

## Average Debt-to-Income Ratio

```python
dti_ratio = df ['dti'].mean()
MTD_dti_ratio = df[df['Issue_Month'] == 12]['dti'].mean()
PMTD_dti_ratio = df[df['Issue_Month'] == 11]['dti'].mean()
MoM_dti_ratio = round(((MTD_dti_ratio - PMTD_dti_ratio) / PMTD_dti_ratio)*100, 2)
```

```python
print(f'Average Debt to Income Ratio: {dti_ratio:,.2f}%')
print(f'MTD Average Debt to Income Ratio: {MTD_dti_ratio:,.2f}%')
print(f'PMTD Average Debt to Income Ratio: {PMTD_dti_ratio:,.2f}%')
print(f'MoM Average Debt to Income Ratio: {MoM_dti_ratio}%')
```

 ## Good Loan ad Bad Loan

```python
# Seperate good loan and Bad loan data
good_loan_df = df[df['loan_status'].isin(['Fully Paid', 'Current'])]
bad_loan_df = df[df['loan_status'] == 'Charged Off']
```

## Good Loan KPI

### Total Good Loan Applicants

```python
total_good_loan_applicants = good_loan_df['id'].count()
good_loan_applicants_pct = (total_good_loan_applicants/total_applicants)*100
good_loan_funded_amount = good_loan_df['loan_amount'].sum()
good_loan_received_amount = good_loan_df['total_payment'].sum()
```

```python
print(f'Total Good Loan Applicants: {total_good_loan_applicants:,}')
print(f'Total Good Loan Applicants %: {good_loan_applicants_pct:,.2f}%')
print(f'Total Good Loan Funded Amount: ${good_loan_funded_amount:,.2f}')
print(f'Total Good Received Amount: ${good_loan_received_amount:,.2f}')
```

## Total Bad Loan Applicants

```python
total_bad_loan_applicants = bad_loan_df['id'].count()
bad_loan_applicants_pct = (total_bad_loan_applicants/total_applicants)*100
bad_loan_funded_amount = bad_loan_df['loan_amount'].sum()
bad_loan_received_amount = bad_loan_df['total_payment'].sum()
```

```python
print(f'Total Bad Loan Applicants: {total_bad_loan_applicants:,}')
print(f'Total Bad Loan Applicants %: {bad_loan_applicants_pct:,.2f}%')
print(f'Total Bad Loan Funded Amount: ${bad_loan_funded_amount:,.2f}')
print(f'Total Bad Received Amount: ${bad_loan_received_amount:,.2f}')
```

## Loan Status Grid View

```python
# Currency Converter
def currencyConverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}''{:,.2f)'.format)

# Percentage Converter
def pctConverter(dataframe, suffix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map('{:,.2f}'f'{suffix}'.format)
```

```python
MTD_fund = df[df['Issue_Month'] == 12]
data = MTD_fund.groupby('loan_status')['loan_amount'].sum().reset_index()
received = MTD_fund.groupby('loan_status')['total_payment'].sum().reset_index()
received
```

```python
loan_grid = df.groupby('loan_status').agg(
    Total_Loan_Applicants = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Amount_Received = ('total_payment', 'sum'),
    Average_Interest_Rate = ('int_rate', 'mean'),
    Average_DTI = ('dti', 'mean')
).reset_index().rename(columns = {'loan status': 'Loan Status'})

# Merging analyzed data columns
loan_grid['MTD_Funded_Amount'] = data['loan_amount']
loan_grid['MTD_Received_Amount'] = received['total_payment']

# replacing unwanted value in column names
loan_grid.columns = loan_grid.columns.str.replace('_', ' ', regex= False)

# using functions
currencyConverter(loan_grid, '$', 'Total Funded Amount', 'Total Amount Received', 'MTD Funded Amount', 'MTD Received Amount')
pctConverter(loan_grid, '%', 'Average Interest Rate', 'Average DTI')
```
