```python
'''
1 Year business
Average Profit = 500000 -> Sample

Total Sales = 1300000 -> Population

Types of Hypothesis Testing
------------------------------------
1. T Test
    a. One Sample T Test
    b. Two Sample T Test

2. ANOVA (Analysis of Variance) Test

- Statistics
- Probability

Stats -> (Actual mean - Sample mean) / (sd / root(total_number_of_data))

- If stats value increases P-value decreases
- If stats value decreases P-value increases
'''
```

```python
# Statistical calculation -> scipy -> scientific python
from scipy import stats
import pandas as pd
```

```python
df = pd.read_csv('dataset/pizza_sales.csv')
df.head()
```

## One Sample T Test

### Is the average total price of all pizza sales different from $15?


```python
alpha = 0.05
sample = 15
population = df['total_price']
```

```python
'''
import numpy as np
actual_mean = population.mean()
sd = population.std()
total_data = np.sqrt(48619)

top_val = actual_mean - sample
bot_val = sd / total_data

top_val / bot_val
'''
```

```python
actual_mean
```

```python
# stats.ttest_1samp(population, popmean=sample)
t_statistics, p_value = stats.ttest_1samp(population, popmean=sample)
```

```python
ho = f'The average total price of aall pizza sales equals to ${sample}.'
ha = f'The average total price of all pizza sales is not equal to ${sample}.'
print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f'Decision: Failed to accept null hypothesis.\n Conclusion: Needs to accept alternate hypothesis.\nHa: {ha}')
else:
    print(f'Decision: Failed to reject null hypothesis.\nConclusion: Needs to accept null hypothesis.\nHo: {ho}')
    
```

## Two Sample T Test

### The average unit price of Classic Pizzas and Chicken Pizzas same?

```python
alpha = 0.05
classic_price = df[df['pizza_category'] == 'Classic']['unit_price']
chicken_price = df[df['pizza_category'] == 'Chicken']['unit_price']
```

```python
# stats.ttest_ind(category1, category2, equal_var=True/False)
t_statistics, p_value = stats.ttest_ind(classic_price, chicken_price, equal_var=False)
```

```python
ho = 'The average unit price of Classic Pizzas and Chicken Pizzas are equal.'
ha = 'The average unit price of Classic Pizzas and Chicken Pizzas are not equal.'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f'Decision: Failed to accept null hypothesis.\nConclusion: Needs to accept alternate hypothesis.\nHa: {ha}')
else:
    print(f'Decision: Failed to reject null hypothesis.\nConclusion: Needs to accept null hypothesis.\nHo: {ho}')
```

# ANOVA Test(Analysis of Variance)

## Do the average total_price generated by all four pizza categories(Classic, Chicken, Supreme, and Veggie) differ significantly from one another?

```python
alpha = 0.05
categories = ['Classic', 'Chicken', 'Supreme', 'Veggie']
grouped_prices = []

for category in categories:
    price = df[df['pizza_category'] == category]['total_price']
    grouped_prices.append(price)
```

```python
t_statistics, p_value = stats.f_oneway(*grouped_prices)
```

```python
pizza_price = df.groupby('pizza_category')['total_price'].mean().rename_axis('Pizza Category').rename(
    'Total Price'
).map("${:,.2f}".format).to_markdown(numalign='center', stralign='center')
```

```python
ho = 'The average total price generated by all four pizza categories are equal to one another.'
ha = 'The average total price generated by all four pizza categories are not equal to one another .'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
print('-'*38)
print(pizza_price)
print('-'*36)
```

```python
if p_value < alpha:
    print(f'Decision: Failed to accept null hypothesis.\nConclusion: Needs to accept alternate hypothesis.\nHa: {ha}')
else:
    print(f'Decision: Failed to reject null hypothesis.\nConclusion: Needs to accept null hypothesis.\nHo: {ho}')
```

# Chi Squared Test

## Is Pizza Category and Pizza Size are dependent to one another? 

```python
alpha = 0.05
# contingency analysis -> crosstab (Cross Tabulation)
contingency = pd.crosstab(df['pizza_category'], df['pizza_size'])
contingency
```

```python
# dof = (number of rows - 1) * (number of columns - 1)
# dof = (4 - 1) * (5 - 1)
# dof
```

```python
t_statistics, p_value, dof, ef = stats.chi2_contingency(contingency)
```

```python
ho = 'Pizza Category and Pizza size are dependent to one another.'
ha = 'Pizza Category and Pizza size are independent to one another .'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f'Decision: Failed to accept null hypothesis.\nConclusion: Needs to accept alternate hypothesis.\nHa: {ha}')
else:
    print(f'Decision: Failed to reject null hypothesis.\nConclusion: Needs to accept null hypothesis.\nHo: {ho}')
```
